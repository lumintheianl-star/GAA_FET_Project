```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
# 1. Khởi tạo mảng nhiệt độ T từ 100K đến 1000K bằng NumPy
T = np.linspace(100, 1000, 500)
k_B = 8.6173e-5      # Hằng số Boltzmann (eV/K)
E_minus_Ef = 0.1     # Mức năng lượng vùng dẫn (eV)
# 2. Công thức Fermi-Dirac tính xác suất chiếm dụng của hạt tải
f_E = 1 / (np.exp(E_minus_Ef / (k_B * T)) + 1)
# 3. Quản lý cấu trúc bảng dữ liệu bằng Pandas
df = pd.DataFrame({"Nhiệt độ T (K)": T, "Xác suất f(E)": f_E})
# 4. Xuất đồ thị trực quan bằng Matplotlib
plt.figure(figsize=(6, 3.5))
plt.plot(df["Nhiệt độ T (K)"], df["Xác suất f(E)"], color="red", linewidth=2)
plt.title("Đồ thị Fermi-Dirac - Tuần 1 Nghiên cứu")
plt.grid(True, linestyle="--")
plt.show()
# Hiển thị 5 dòng đầu của bảng dữ liệu Pandas
df.head()
```
